```python
import re
import pandas as pd
import json
```

```python
data = {}

with open ('input-data/shortcodes.md') as f:
    f=f.read()
    for line in f.strip().splitlines():
        line=line.strip()
        if not line:
            continue
        if line.endswith(":") and not line.strip().startswith("{{"):
            path=line[:-1]
            data[path] = []
        else:
            full = re.match(r"(\d+):\s*(.+)", line)
            sc = re.search(r"(?<=\{\{).*?(?=\}\})", line)
            if sc and full and path:
                sc_type = sc.group(0).split(" ")[1]
                if not sc_type.startswith("/"):
                    line_number = int(full.group(1))
                    fullcontent = full.group(2).strip()
                    data[path].append({
                        "line": line_number,
                        "shortcode_type": sc_type,
                        "shortcode_full": fullcontent
                    })

jsonOut = json.dumps(data,indent=2)

with open('dumps/shortcodes-dump.json', 'w') as out:
    json.dump(data, out, ensure_ascii=False,)
```

```python
with open ('dumps/shortcodes-dump.json', 'r') as f:
    d = json.load(f)

with open('dumps/shortcodes-dump.csv', 'w', newline='', encoding='utf-8') as csvfile:
    fields = ['path', 'line', 'shortcode_type', 'shortcode_full']
    writer = csv.DictWriter(csvfile, fieldnames=fields)

    writer.writeheader()

    for filepath, entries in d.items():
        for entry in entries:
            print(entry)
            writer.writerow({
                'path': filepath,
                'line': entry['line'],
                'shortcode_type': entry['shortcode_type'],
                'shortcode_full': entry['shortcode_full']
            })
```
